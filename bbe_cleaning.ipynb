## Dropping Columns ##
<p> Taking a look at the above list, there are several that I won't be using in this juvinile analysis. I'm going to go ahead and drop them below. <p>

```python
columns_to_drop = ["characters", "description", "price", "coverImg", "setting", "series", "bookFormat", "firstPublishDate", "edition"]
df.drop(columns = columns_to_drop, axis=1, inplace=True)
df.columns
```

## Splitting a column to more data ##

<p> Even though I just cleared out several rows, there is one data set "ratingsByStars" that I am interested in breaking apart for comparrison. So I am goingto split them into a new set and concat it back into the df. Then drop the original column. <p>

```python
new_columns = df["ratingsByStars"].str.split(",", expand=True)
new_columns.columns = ["5_Stars", "4_Stars", "3_Stars", "2_Stars", "1_Star"]
df = pd.concat([df,new_columns], axis=1)
df.drop(columns = "ratingsByStars", inplace=True)
df.head()
```

## Stripping some special characters ##
<p> Now these numerical dataframes have special characters that I don't want, so I am just going to strip them out one at a time. I forget if there was a way to do this or if it really even matters, but I am doing it anyways.<p>

```python
df["5_Stars"] = df["5_Stars"].str.strip("[")
df["5_Stars"] = df["5_Stars"].str.strip("'")
df["4_Stars"] = df["4_Stars"].str.strip("'")
df["3_Stars"] = df["3_Stars"].str.strip("'")
df["2_Stars"] = df["2_Stars"].str.strip("'")
df["1_Star"] = df["1_Star"].str.strip("'")
df["1_Star"] = df["1_Star"].str.strip("]")
df.head(2)
```

## Roadblock ##

<p> my only roadblock came in trying to clean up the new columns I created, it would not deleted both opening and closing ' <p>

## null percentage ##
Next I want to run and see how many nulls are left in the remaining df and clean up those remaining frames.

```python
df.isnull().sum() * 100 / len(df)
```

<p> With the remaining rows that I have to work with, I'm going to startgoing through to take care of the null values. Language has the largest %, and since this is categorical, to take care of null values I will fill it in with "Unknown" so it has a placeholder. This takes care of its percentage of nulls.

```python
df["language"].unique()
df["language"].fillna("Unknown", inplace=True)
df.isnull().sum() * 100 / len(df)
```

<p> Note: some languages have additional information basing on the time period. EX: old middle english. See about removing the comma so that it reads all as one category.<p>

## Do the same above but for publisher ##

```python
df["publisher"].unique()
df["publisher"].fillna("Unknown", inplace=True)
df.isnull().sum()
```

## Duplicate rows ##

<p> To take care of some duplicated rows, I'm going to run the loc command to see how many there are. There appear to be over 100, so I am going to keep the first instance and delete the rest. <p>

```python
df.loc[df.duplicated(keep=False)]
df.drop_duplicates(keep='first', inplace=True)
```

## Rows with null values ##

<p> I decided with the remaining values, that there were going to be a significant gap to keep, so for any rows with empty rows in the below listed subset, I am just going to proceed and drop. <p>

```python
df.dropna(subset=["likedPercent", "pages", "4_Stars", "3_Stars", "2_Stars", "1_Star", "publishDate"], inplace=True)
df.isnull().sum()
```

## Fixed ISBN ##

<p> upon review, I realized that isbn had multiple values listed for it. Since this is not an int, I am going to fix it to say that it is an unknown isbn <p>


```python
df["isbn"].value_counts()
```

```python
for i in df[df["isbn"]== "9999999999999"].index:
    df.loc[i, "isbn"] = "unknown isbn"
```

## Creating visuals ##

<p> This cleaning process has satisfied me for the day. Enough for me to create some visuals.

```python
pop_vote = df.groupby("author")["numRatings"].max().sort_values(ascending=False)
top_13_values = pop_vote[:13]
top_13_values
```

```python
plt.bar(top_13_values.index, top_13_values.values)
plt.show()
```

## Initial Chart creation ##

<p> Oh no! This chart is awful display to not be legible and unknown axis. I will fix it up<p>

```python
plt.figure(figsize=(14, 10))
plt.bar(top_13_values.index, top_13_values.values)

ax= plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.xlabel("Authors", size = 6, color = "black")
plt.xticks(rotation=45, ha='right', wrap=True)
plt.ylabel("Numer of Ratings Recieved (millions)", size=10, color="black")
plt.title("Highest Rated Authors From Best Books Ever", fontsize= 16, pad=20, color="black")



plt.show()
```
